```python
import pandas as pd
import numpy as np
df=pd.read_csv("quikr_car (2).csv")
```

```python
df.head()
```

```python
df["Price"] = df["Price"].replace("Ask For Price", np.nan)
df["Price"] = df["Price"].str.replace(",", "", regex=False)
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
```

```python
df = df.dropna(subset=["Price"])
df["Price"] = df["Price"].astype(int)
```

```python
df["Price"].unique()
```

```python
df.info()
```

```python
df["kms_driven"] = (
    df["kms_driven"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("kms", "", case=False, regex=False)
    .str.strip()
)

df["kms_driven"] = pd.to_numeric(df["kms_driven"], errors="coerce")
```

```python
df = df.dropna(subset=["kms_driven"])
```

```python
df["kms_driven"] = df["kms_driven"].astype("Int64")
```

```python
df[df["kms_driven"].isin(["Petrol", "Diesel"])]
```

```python
df['kms_driven'].unique()
```

```python
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df["year"] = df["year"].astype("Int64")
```

```python
df["year"].unique()
```

```python
df["company"].unique()
```

```python
df["name"]=df["name"].str.split(" ").str.slice(0,3)
```

```python
df["name"]=df["name"].str.join(' ')
```

```python
df=df.reset_index(drop=True)
```

```python
df=df[df["Price"]<6e6].reset_index(drop=True)
```

```python
df.describe()
```

```python
df.to_csv('cleaned_car.csv')
```

MODEL

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder
```

```python
x = df.drop("Price", axis=1)
y = df["Price"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
coln_trans = make_column_transformer(
    (OneHotEncoder(handle_unknown="ignore"),
     ["name", "company", "fuel_type"]),
    remainder="passthrough"
)
```

```python
lr=LinearRegression()
```

```python
pipeline=make_pipeline(coln_trans,lr)
```

```python
pipeline.fit(x_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = pipeline.predict(x_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    random_state=0
)

rf_pipeline = make_pipeline(coln_trans, rf)

rf_pipeline.fit(x_train, y_train)

rf_pred = rf_pipeline.predict(x_test)

print("Random Forest MAE:",
      mean_absolute_error(y_test, rf_pred))

print("Random Forest RMSE:",
      np.sqrt(mean_squared_error(y_test, rf_pred)))

print("Random Forest R2:",
      r2_score(y_test, rf_pred))
```
